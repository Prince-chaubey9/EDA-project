```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
data= pd.read_csv('Sample - Superstore.csv',encoding='cp1252')
```

```python
data= pd.DataFrame(data)
data.head(10)
```

```python
data.info()
```

```python
data['Order Date']= pd.to_datetime(data['Order Date'])
```

```python
# adding new column contain month name and year 

data['month']= data['Order Date'].dt.month_name()
data['year']=data['Order Date'].dt.year
data.head()
```

# monthly sale 

```python
monthly_sale= data.groupby('month')['Sales'].sum().reset_index()
monthly_sale.sort_values('Sales',inplace=True)
monthly_sale
```

```python
# get a bar plot of monthly sale 
plt.figure(figsize=(15, 10))  # define dimension of chart
plt.bar(monthly_sale['month'],monthly_sale['Sales'])
plt.xlabel('MONTH')
plt.ylabel('SALE IN THOUSAND')
plt.title('MONTHLY SALE CHART ')
plt.show()
plt.tight_layout()
```

# category sale 

```python
category_sale= data.groupby('Category')['Sales'].sum().reset_index()
category_sale.sort_values('Sales',inplace=True)
category_sale
```

```python
# percentage sale of category of company 
plt.pie(category_sale['Sales'],labels= category_sale['Category'],autopct='%1.1f%%')
```

```python
plt.figure(figsize=(4, 4))
clrs=['red','green','blue','pink']
plt.bar(category_sale['Category'],category_sale['Sales'],color=clrs)
plt.xlabel('category',fontweight='bold')
plt.ylabel('sale')
plt.title(' category vs sale', fontweight='bold',fontsize=15)
```

# monthly profit

```python
monthly_profit= data.groupby('month')['Profit'].sum().reset_index()
monthly_profit.sort_values('Profit',inplace=True)
monthly_profit
```

```python
# get a bar plot of monthly sale 
plt.figure(figsize=(15, 10))  # define dimension of chart
plt.bar(monthly_profit['month'],monthly_profit['Profit'])
plt.xlabel('MONTH')
plt.ylabel('PROFIT IN THOUSAND')
plt.title('MONTHLY PROFIT CHART ',fontweight='bold')
plt.show()
plt.tight_layout()
```

```python
plt.pie(monthly_profit['Profit'],labels=monthly_profit['month'],autopct='%1.1f%%')
```

# month;y sales to profit ratio

```python
monthly_ratio= data.groupby('month')[['Sales','Profit']].sum().reset_index()
monthly_ratio['ratio']= monthly_ratio['Sales']/monthly_ratio['Profit']
monthly_ratio.sort_values('ratio',inplace=True)
monthly_ratio
```

```python
plt.figure(figsize=(10, 5))
bar=plt.bar(monthly_ratio['month'],monthly_ratio['ratio']) # chart ko bar veriable m store kia 
plt.bar_label(bar) # use bar_label fun m pass kiya to values uper show ho gyi 
plt.xlabel('MONTH')
plt.ylabel('RATIO')
plt.title(' sales to profit ratio comparision of months')
```
